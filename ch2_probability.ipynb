# Probability
Probability is the theoretical study of measuring certainty that an event will happen. It is a foundational discipline for statistics, hypothesis testing, machine learning, etc.

* **Scale:** Probability $P(X)$ must strictly fall between $0.0$ and $1.0$ (or $0\%$ to $100\%$).
* **Complement:** The probability of an event *not* happening is $P(\text{not } X) = 1 - P(X)$.
* **Sum of Outcomes:** The probabilities of all mutually exclusive outcomes must sum to $1.0$.

### Probability vs Likelihood
* **Probability** is about quantifying predictions of events yet to happen.
* **Likelihood** is measuring the frequency of events that already occurred.

### Odds vs. Probability
Odds $O(X)$ express how much more likely an event is to happen than *not* to happen (e.g., an odds of $2.0$ means an event is twice as likely to occur as not).

* **Converting Odds to Probability:**
  $$P(X) = \frac{O(X)}{1 + O(X)}$$
* **Converting Probability to Odds:**
  $$O(X) = \frac{P(X)}{1 - P(X)}$$

**Why Odds Matter:** Odds are often more intuitive for subjective beliefs and serve as the foundation for **Bayesian statistics** (Bayes Factor) and **Logistic Regression** (log-odds).

### Probability vs. Statistics
**Probability** is purely theoretical of how likely an event is to happen and does not require data.
**Statistics**, on the other hand, cannot exist without data and uses it to discover probability and provides tools to describe data.

## Probability Math
When we work with a single probability of an event $P(X)$, known as *marginal probability*.

### Joint Probabilities
A joint probability measures the likelihood of two or more events occurring simultaneously. It act as a logical $AND$ operator.
We are looking for the probability that event $A$ happens $AND$ Event $B$ happens
$$P(A \text{ AND } B) = P(A) \times P(B)$$

For example, I want to find the probability of flipping a heads $AND$ rolling a 6.
$$P(\text{heads}) = \frac{1}{2}$$

$$P(6) = \frac{1}{6}$$

$$P(\text{heads AND } 6) = \frac{1}{2} \times \frac{1}{6} = \frac{1}{12} \approx .08333$$

---
### Union Probability
A union probability qualifies the likelihood of Event A $OR$ Event B occurring. It's denoted as $P(A \cup B)$ or $P(A \text{ or } B)$.

#### 1. Mutually Exclusive Events
Mutually exclusive events cannot happen at the exact same time (e.g., rolling a single die and getting both a $4$ and a $6$).
For mutually exclusive events, simply sum their individual probabilities:
$$P(A \text{ or } B) = P(A) + P(B)$$

For example:
$$P(4 \text{ or } 6) = P(4) + P(6) = \frac{1}{6} + \frac{1}{6} = \frac{2}{6} = \frac{1}{3} \approx 0.3333$$

#### 2. Non-Mutually Exclusive Events
Non-mutually exclusive events can happen simultaneously (e.g., flipping Heads on a coin AND rolling a $6$ on a die).

Simply adding their individual probabilities creates a double-counting error by counting the joint outcome ($H6$) twice. To correct this, subtract the joint probability $P(A \text{ and } B)$. This is known as the **Sum Rule of Probability**:

$$P(A \text{ or } B) = P(A) + P(B) - P(A \text{ AND } B)$$

For example:
$P(\text{Heads}) = \frac{1}{2}$
$P(6) = \frac{1}{6}$
$$P(A \text{ or } B) = P(A) + P(B) - P(A) \times P(B)$$

### Conditional Probability and Bayes' Theorem
Is the probability of a event $A$ occurring given event $B$ has occurred. It is typically expressed as $P(A \text{ GIVEN } B)$ or $P(A|B)$.

For example, we had the percentage of people diagnosed with cancer ($0.5\%$) and the percentage of people who drink coffee ($65\%.$).
$$P(\text{ Coffee }) = .65$$
$$P(\text{ Cancer }) = .005$$
$$P(\text{ Coffee|Cancer }) = .85$$

There's a powerful little formula called Bayes'Theorem, and we can use it to flip conditional probabilities:
$$P(A|B) = \frac{P(B|A)P(A)}{P(B)}$$

If we plug the information we already have into this formula, we can solve for the probability someone has cancer given they drink coffee:

$$P(\text{Cancer}|\text{Coffee}) = \frac{P(\text{Coffee}|\text{Cancer}) * P(\text{Coffee})}{P(\text{Cancer})}$$

$$P(\text{Cancer}|\text{Coffee}) = \frac{.85 * .005}{.65} = .0065$$

```python
p_coffee_drinker = 0.65
p_cancer = 0.005
p_coffee_drinker_given_cancer = 0.85

p_cancer_given_coffee_drinker = p_coffee_drinker_given_cancer * p_cancer / p_coffee_drinker

print(p_cancer_given_coffee_drinker)
```

### Joint and Union Conditional Probabilities

#### Joint Probability with Dependence
When events are dependent, use the conditional probability rather than the baseline prior probability:

$$P(A \text{ and } B) = P(A|B) \times P(B)$$

#### Example

$$\text{Calculation 1: } P(\text{Coffee and Cancer}) = P(\text{Coffee}|\text{Cancer}) \times P(\text{Cancer}) = 0.85 \times 0.005 = 0.00425$$

$$\text{Calculation 2: } P(\text{Coffee and Cancer}) = P(\text{Cancer}|\text{Coffee}) \times P(\text{Coffee}) = 0.0065 \times 0.65 = 0.00425$$

---

#### General Sum Rule (Union Probability)
To calculate $P(A \text{ or } B)$ when events are not independent, subtract the updated joint probability to prevent double-counting:

$$P(A \text{ or } B) = P(A) + P(B) - P(A|B) \times P(B)$$

*Note:* For mutually exclusive events, $P(A|B) \times P(B) = 0$, which simplifies the equation back to $P(A) + P(B)$.

## Binomial Distribution
Binomial distribution measures how likely $k$ successes can happen out of $n$ trials given $p$ probability.
$$
P(X=k) = \binom{n}{k} p^k (1-p)^{n-k}
$$


$$
\binom{n}{k} = \frac{n!}{k!(n-k)!}
$$



```python
n = 10
p = 0.9

for k in range(n + 1):
    probability = binom.pmf(k, n, p)
    print("{0} - {1}".format(k,probability))
```

```python
# Binomial Distribution from scratch
# Define factorial function
def factorial(n):
    f = 1
    for i in range(n):
        f *= (i + 1)
    return f

# Generates the coefficient needed for the binomial distribution
def binomial_coefficient(n, k):
    return factorial(n) / (factorial(k) * factorial(n -k))

# Binomial distribution calculates the probability of k events out of n trials given the p probability of k occurring
def binomial_distribution(k, n, p):
    return binomial_coefficient(n, k) * (p ** k) * (1.0 - p) ** (n - k)

# 10 trials where each has 90% success probability
n = 10
p = 0.9

for k in range(n + 1):
    probability = binomial_distribution(k, n, p)
    print("{0} - {1}".format(k, probability))
```

## Beta Distribution
Beta distributions allows us to see the likelihood of different underlying probabilities for an event to occur given $alpha$ successes and $beta$ failures.

```python
a = 8
b = 2

p = beta.cdf(.90, a, b)

print(p)
```

```python
# Subtracting to get a right area in a beta distribution
a = 8
b = 2

p = 1.0 - beta.cdf(.90, a, b)

print(p)
```

```python
# Beta Distribution from scratch
# Define factorial function
def factorial(n):
    f = 1
    for i in range(n):
        f *= (i + 1)
    return f

def approximate_integral(a, b, n, f):
    delta_x = (b - a) / n
    total_sum = 0

    for i in range(1, n + 1):
        midpoint = 0.5 * (2 * a + delta_x * (2 * i - 1))
        total_sum += f(midpoint)
    return total_sum * delta_x

def beta_distribution(x: float, alpha: float, beta: float) -> float:
    if x < 0.0 or x > 1.0:
        raise ValueError("x must be between 0.0 and 1.0")
    numerator = x ** (alpha - 1.0) * (1.0 - x) ** (beta - 1.0)
    denominator = (1.0 * factorial(alpha - 1) * factorial(beta - 1)) / (1.0 * factorial(alpha + beta - 1))

    return numerator / denominator

greater_than_90 = approximate_integral(a=.90, b=1.0, n=1000, f=lambda x: beta_distribution(x, 8, 2))
less_than_90 = 1.0 - greater_than_90

print("GREATER THAN 90%: {}, LESS THAN 90%: {}".format(greater_than_90,less_than_90))
```
